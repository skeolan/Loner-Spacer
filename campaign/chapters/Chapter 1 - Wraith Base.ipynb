# WRAITH BASE



**Docking Berth**  

**3500-08-03 · 11:42 Station Time**  

***Horizon Chaser* — Cockpit**



> *WRAITH BASE NEWS FEED — The* Argent Surveyor *has become the third survey vessel lost in the Charybdis Rift. Concordat Survey Command presumes all thirty-one souls aboard dead. No recovery signal has been authenticated, and the destructive agent remains unidentified. Civilian traffic is advised to remain outside the revised hazard boundary.*

## GM

Beyond the station's scarred hull, the Charybdis Rift twists the stars into an ugly knot.

Inside the *Horizon Chaser*, the morning is smaller and more immediate. You are at the cockpit worktable finishing final preflight, one hand moving down the checklist on the primary touchscreen while the ship quietly verifies each system behind you.

You filed the clearance request first thing this morning. Port Authority takes for-fucking-*ever* to approve an independent nobody, and there was no point waiting to solve the fuel problem before starting that clock.

The cargo manifest is shorter than the checklist: one encrypted Cinder Ledger data package secured inside the smuggler's hold, nothing else aboard worth selling, and seventy-two hours left before your next payment comes due.

Preflight is otherwise finished. Your next job is to leave the cockpit, work Wraith Base's fuel brokers, dock crews, and less reputable channels, and somehow scrounge enough hyperfuel to make the departure math work. After that, you can finish the courier run already hanging over you and put distance between the *Chaser* and anyone inclined to ask what she is carrying.

Before you can close the checklist, the right-hand communications display sounds a discreet priority chime.

## Player

I wipe the manifest from the main display before accepting the channel. Le Goff gets my face, the back of the pilot's chair, and nothing else.

> [!QUOTE] **NOVA:**
> Commander. Concordat liaison doesn't usually call a broke independent courier to improve her morning.

I glance at the priority seal, then back to Le Goff.

> [!QUOTE] **NOVA:**
> I'm already carrying a contract, and the clock on it isn't friendly. If this is another job, make it quick. If it isn't, make it *quicker*.

## GM

Le Goff appears in a dark Concordat uniform, silver at the collar, the liaison office behind her reduced to glass reflections and disciplined shadow. Her eyes move once across the telemetry your ship shares with the authenticated channel: fuel state, vessel class, navigation certification.

> [!QUOTE] **LE GOFF:**
> The *Argent Surveyor* ejected its navigation core before it was destroyed. That core recorded the attack, but its beacon is buried in the Rift's distortion. Your navigation suite can find it. Bring it back to Wraith Base. I'm transmitting the specifics to you now.

There is no greeting and no attempt to make the work sound safe. An authenticated contract packet opens on the starboard display.

> [!QUESTION] **STARBOARD COMMS DISPLAY // CONTRACT PROPOSAL**
> ```ini
> [ASSIGNMENT]
> issuer = CONCORDAT
> objective = RECOVER ARGENT SURVEYOR NAVIGATION CORE
> location = CHARYBDIS RIFT
>
> [CONSIDERATION]
> hyperfuel = FULL REFUEL
> hardware = MILITARY-GRADE ENERGY CORE
> departure_clearance = ARRANGED
>
> [RESPONSE]
> status = PENDING
> action = ACCEPT | DECLINE
> ```

## Player

I lean back, studying the Commander as though the offer itself might be carrying contraband.

> [!QUOTE] **NOVA:**
> Refuel, hardware, and suddenly my clearance stops being a problem. That's generous enough to make me nervous. What's the core really worth to the Concordat? And why me?

I gesture around the cramped courier cockpit.

> [!QUOTE] **NOVA:**
> The *Chaser* moves packages. She doesn't hunt whatever tears survey ships apart. And I don't usually see security types like you as customers so much as complications.

I keep my gaze on Le Goff.

> [!QUOTE] **NOVA:**
> So try the part you left out.

## GM

> [!QUOTE] **LE GOFF:**
> Because the *Horizon Chaser* is small, reliable, and carrying an advanced navigation suite precise enough to follow that beacon through the distortion. Because you're a navigation ace. And because the Concordat needs the core recovered before anything else comes out of the Rift.

Le Goff lets the answer settle, then repeats the terms from the contract packet aloud.

> [!QUOTE] **LE GOFF:**
> We refuel your ship. We install a military-grade energy core when you return. And I can see that you receive departure clearance.

The last phrase has barely crossed the channel when the cockpit lights gutter under a berth-power override. On the *Chaser*'s status board, the ordinary green clamp indicators turn red.

> [!DANGER] **PRIMARY TOUCHSCREEN // BERTH CONTROL**
> ```diff
> -- [!] BERTH POWER OVERRIDE ........ ACTIVE
> -- [!] EXTERNAL CLAMPS ............. SECURITY HARD LOCK
> -- [!] DEPARTURE INTERLOCK ......... ENGAGED
> ```

Metal drives against metal through the hull. The deck shudders beneath your boots as Wraith Base tightens its grip. A second transmission begins forcing its way onto the display.

For one unguarded instant, before the new channel covers half her face, Le Goff smirks.

## Player

My chair snaps backward into the console rail as I come halfway to my feet. I catch the edge of the worktable with one hand; the other flies across the berth controls, checking power, seals, and clamp telemetry in quick succession.

Then I see Le Goff's smirk.

> [!QUOTE] **NOVA:**
> You knew.

Le Goff draws breath.

> [!QUOTE] **LE GOFF:**
> What I know is—

I slap **HOLD**. The Commander's voice cuts off mid-sentence, her image collapsing into a silent amber icon while the Port Authority hail takes over the cockpit.

## GM

A compulsory Port Authority channel forces itself open beside the suspended Concordat transmission. Marshal Dafne Catalina Castrillo Guitart appears on-screen in a severe station-services office, two armed officers visible over her shoulders.

> [!QUOTE] **CASTRILLO:**
> Nova Darkreach, this is Marshal Castrillo. Acknowledge.

The command carries the flat confidence of someone accustomed to doors opening before she reaches them.

## Player

I answer the Port Authority hail and fold my arms before the channel finishes resolving.

> [!QUOTE] **NOVA:**
> Marshal. Locking down my ship before you say hello is a hell of an introduction. Under whose authority, and on what grounds? Send me the order before anyone touches my airlock.

My eyes flick once toward the compartment-status panel. The smuggler's hold still reports sealed.

## GM

Castrillo's expression does not move.

> [!QUOTE] **CASTRILLO:**
> Your departure clearance is suspended. We have credible intelligence that your vessel contains an undeclared compartment. Open your airlock for inspection.

She offers nothing more. The exterior pickup catches the measured impact of boots entering the docking collar.

## Player

I keep my voice level and give Castrillo nothing useful to hear.

> [!QUOTE] **NOVA:**
> Transmit the inspection order and its scope to my ship. You and your people can wait outside while I verify it.

I touch **RECEIVE DOCUMENTS**, not **OPEN AIRLOCK**.

## GM

Your request registers. An authenticated packet lands on the starboard display almost immediately—too quickly to have been drafted in response.

> [!WARNING] **STARBOARD COMMS DISPLAY // INSPECTION ORDER**
> ```ini
> [AUTHORITY]
> issuer = WRAITH BASE PORT AUTHORITY
> officer = MARSHAL DAFNE CATALINA CASTRILLO GUITART
> authentication = VERIFIED
>
> [ORDER]
> vessel = HORIZON CHASER
> basis = SUSPECTED UNDECLARED COMPARTMENT
> scope = HULL-COMPARTMENT VERIFICATION
> status = ACTIVE
> ```

The signature chain is valid. The order says nothing about who supplied the intelligence.

Beyond the inner airlock, the measured footsteps stop. Three Port Authority transponders hold position in the docking collar. The *Chaser* still controls her own hatch, but Wraith Base controls the clamps around her hull.

> [!QUOTE] **CASTRILLO:**
> You have your order. Open the airlock. If you continue to delay, my officers will execute it without your cooperation.

Le Goff's channel remains a silent amber **HOLD** icon at the edge of the display. Castrillo waits on the live screen.

## Player

I clench my jaw and, keeping my eyes locked on Castrillo's through the open video-call window, reach out and tap the icon to resume the comm session with Le Goff.

> [!QUOTE] **NOVA:**
> Commander, I accept your offer. But if you want me flying for you, there's a little matter with the Port Authority that needs to be cleaned up. Call off Castrillo and her hounds, or I'm—obviously—not going anywhere. I seem to recall your request had some *urgency*, no?

> Marshal Castrillo, suspend your inspection. Darkreach and the *Horizon Chaser* are now under Concordat contract for an emergency recovery operation. My office accepts responsibility for her departure.

> [!QUOTE] **CASTRILLO:**
> Your office may accept responsibility. It does not erase my order. I will stay execution until the *Horizon Chaser* departs. The inspection resumes the moment she returns to Wraith Base.

> [!QUOTE] **LE GOFF:**
> A Concordat fuel tender is routing to your berth. Once you're fueled, leave. Bring me that core before the Rift claims a fourth ship.

## Player

I smile wryly.

> [!QUOTE] **NOVA:**
> “Stay execution” sounds a little more hardcore than she meant it. I hope. But—the *baseless accusation* that the *Horizon Chaser* is operating an undeclared cargo compartment full of *suspicious contraband* notwithstanding—sure, I'll *happily* submit to a thorough inspection the next time I dock at Wraith Base.

It's obviously just a bureaucratic chest-thump; what *idiot* would return to a station carrying anything suspicious if they *know* the Port Authority is going to enforce an inspection on them as soon as the umbilicals go green? *But, sure. Let Castrillo save face. I'll just swing by the Cinder Ledger drop-point in Drake's Divide before I come back from my fetch-quest.*

> [!QUOTE] **NOVA:**
> I can see that fuel tender moving this way, Commander. Pleasure doing business with you, and don't worry—I've got no intention of putting the *Chaser* on the list of things the Rift ate this week, or ever.

## GM

One corner of Le Goff's mouth almost moves. Almost.

> [!QUOTE] **LE GOFF:**
> See that you don't. The fuel transfer is authorized, and your recovery credentials are in the contract packet. Send me the core's telemetry the moment you establish a clean link. Le Goff out.

Her channel closes. The contract packet remains pinned to the starboard display, its response field changed from **ACCEPTED** to **ACTIVE**.

Outside the cockpit glass, a squat Concordat tender glides along the berth guide-lights. Its maneuvering jets pulse once, then go dark as the service arm takes hold beneath the *Chaser*'s port quarter. A vibration passes through the deck—not the brutal impact of the security clamps, but the familiar mechanical shiver of a fuel coupling seating home.
