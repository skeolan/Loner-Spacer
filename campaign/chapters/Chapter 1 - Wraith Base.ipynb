## Wraith Base

The station hangs beside the Charybdis Rift, an ugly knot of distorted starlight where three survey vessels have disappeared.

Nova is aboard the *Horizon Chaser*, docked and nearly broke. The ship needs hyperfuel. The Cinder Ledger expects its next payment in seventy-two hours.

A secure transmission arrives from Commander Dorothée de Le Goff of the Concordat liaison office:

> "Something is coming out of the Rift. It killed thirty-one people aboard the *Argent Surveyor*. Destroy it, recover its navigation core, and we will install a military-grade energy core in your ship. No questions asked."

Before Nova can answer, the cockpit lights dim. Docking clamps engage around the *Chaser* with a heavy metallic crash.

A second transmission opens without permission. Marshal Dafne Catalina Castrillo Guitart appears on-screen, flanked by armed station officers.

> "Nova Darkreach. Your departure clearance is suspended. We have credible intelligence that your vessel contains an undeclared compartment. Open your airlock for inspection."

The smuggler's hold currently contains nothing illegal. It does, however, contain the Cinder Ledger's encrypted payment records, including enough names to start a small war.

Commander de Le Goff is waiting for an answer. Marshal Castrillo's officers are approaching the airlock. The Rift pulses beyond the cockpit glass.

What does Nova do?
